# Структура проекта SBER_PROJECT_GRUSHINDA

```text
SBER_PROJECT_GRUSHINDA/
├── app/                            # Веб-приложение / API слой
│   ├── templates/                  # HTML-шаблоны для веб-интерфейса
│   │   └── index.html              # Главная страница приложения
│   └── api.py                      # Скрипт API
├── catboost_info/                  # Автоматические логи обучения CatBoost
├── dev/                            # Директория для разработки и анализа данных
│   ├── catboost_info/              # Логи обучения CatBoost для локальных тестов
│   ├── data/                       # Исходные датасеты для анализа
│   │   ├── cat_uniq_data.csv       # Уникальные категории данных
│   │   ├── ga_hits.csv             # Хит-логи Google Analytics
│   │   └── ga_sessions.csv         # Сессии Google Analytics
│   ├── Соответствие action : label (1:m).csv   # Маппинг действий к меткам
│   ├── Соответствие category : action (1:m).csv # Маппинг категорий к действиям
│   ├── Соответствие label : action (1:m).csv   # Маппинг меток к действиям
│   ├── Уникальные значения.csv     # Справочник уникальных значений
│   └── EDA.ipynb                   # Разведочный анализ данных (Jupyter Notebook)
├── models/                         # Модуль машинного обучения и продакшн-моделей
│   ├── __pycache__/                # Кэш
│   │   └── my_functions2model.cpython-314.pyc
│   ├── data/                       # Копия данных или валидационный датасет
│   │   ├── cat_uniq_data.csv
│   │   ├── ga_hits.csv
│   │   └── ga_sessions.csv
│   ├── model_v1.pkl                # Сериализованная обученная модель CatBoost 
│   ├── model.py                    # Скрипт для обучения модели
│   └── my_functions2model.py       # Функции предобработки для модели
├── .gitignore                      # Исключения 
└── readme.ipynb                    # Документация проекта
```
